```python
from wordpack.core import WordFiller


filler = WordFiller("white.jpg")
filler.run(
    text="WordPack",
    font_family="cmmi10",
    font_weight="bold",
    letter_spacing=1.1,  # space between letters
    color_weights={'green': 0.5, 'black': 0.5}, # OpTic greeen and black
    base_size=100,  # number of points to extract
    resolution=10,  # coefficient in smoothing
    padding=0.5,                    # Space between circles (pixels)
    min_radius=0.5,                 # Smallest allowed circle (pixels)             # Set to force uniform circle size
    grid_resolution_divisor=25,     # Higher = finer grid, slower
    patience_before_stop=50,        # Failed attempts before moving on
    mega_circle_threshold=0.5,      # Large circle detection threshold
    ray_cast_epsilon=1e-10,         # Numerical stability constant

    # Layout
    margin_percent=0.05,             # Margin as fraction of image width


    # Animation
    batch_size=40,                  # Circles per display update
    show_outline=True,              # Show dashed glyph outlines
)
```
